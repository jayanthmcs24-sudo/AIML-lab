```python

def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:
            if l1 == "~" + l2 or l2 == "~" + l1:
                return (c1 | c2) - {l1, l2}
    return None


def resolution(kb):
    clauses = [set(c) for c in kb]

    while True:
        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):
                result = resolve(clauses[i], clauses[j])

                if result is not None:
                    if not result:
                        print("Empty clause obtained!")
                        print("Goal is Proved.")
                        return

                    if result not in clauses and result not in new_clauses:
                        new_clauses.append(result)

        if not new_clauses:
            print("Goal cannot be Proved.")
            return

        clauses.extend(new_clauses)


# Predefined Knowledge Base
kb = [
    {"P", "Q"},
    {"~P"},
    {"~Q", "R"},
    {"~R"}
]

print("RESOLUTION IN AI")
print("----------------")
resolution(kb)
```

```python

import random
import math

def cost(board):
    c = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                c += 1
    return c

def simulated_annealing():
    board = list(map(int, input("Enter initial state (8 rows, 1-8): ").split()))
    board = [x - 1 for x in board]

    print("Step 0:", [x + 1 for x in board], "Cost:", cost(board))

    step = 0
    while cost(board) != 0:
        current = board[:]
        temperature = 100

        for _ in range(5000):
            new_board = current[:]
            col = random.randint(0, 7)
            new_board[col] = random.randint(0, 7)

            delta = cost(new_board) - cost(current)

            if delta <= 0 or random.random() < math.exp(-delta / temperature):
                current = new_board

            temperature *= 0.995

            if cost(current) == 0:
                break

        board = current
        step += 1

        if step <= 4:
            print("Step", step, ":", [x + 1 for x in board],
                  "Cost:", cost(board))

        if cost(board) != 0:
            board = [random.randint(0, 7) for _ in range(8)]

    print("Solution:", [x + 1 for x in board])
    print("Cost:", cost(board))

    for row in range(8):
        print(" ".join("Q" if board[col] == row else "." for col in range(8)))

simulated_annealing()
```

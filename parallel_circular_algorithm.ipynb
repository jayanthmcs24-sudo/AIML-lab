```python

import numpy as np

# Objective function
def fitness(x):
    return x**2 - 4*x + 4


# Initialize parameters
GRID_SIZE = 10
ITERATIONS = 100
LOWER_BOUND = -10
UPPER_BOUND = 10

# Initialize population randomly
cells = np.random.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    (GRID_SIZE, GRID_SIZE)
)

# Evaluate initial fitness
fitness_values = fitness(cells)

# Store global best solution
best_index = np.unravel_index(
    np.argmin(fitness_values),
    fitness_values.shape
)

best_x = cells[best_index]
best_fitness = fitness(best_x)


# Cellular algorithm
for iteration in range(ITERATIONS):

    new_cells = cells.copy()

    # Visit every cell
    for i in range(GRID_SIZE):
        for j in range(GRID_SIZE):

            # Get 3x3 neighborhood (excluding boundaries)
            neighbors = []

            for di in range(-1, 2):
                for dj in range(-1, 2):

                    ni = i + di
                    nj = j + dj

                    # Check valid grid position
                    if (0 <= ni < GRID_SIZE and
                        0 <= nj < GRID_SIZE):

                        neighbors.append(cells[ni, nj])

            # Find neighbor with minimum fitness
            neighbors = np.array(neighbors)

            best_neighbor = neighbors[
                np.argmin(fitness(neighbors))
            ]

            # Update cell using average
            new_cells[i, j] = (
                cells[i, j] + best_neighbor
            ) / 2

    # Replace old population
    cells = np.clip(
        new_cells,
        LOWER_BOUND,
        UPPER_BOUND
    )

    # Evaluate new fitness
    fitness_values = fitness(cells)

    # Find current best cell
    current_index = np.unravel_index(
        np.argmin(fitness_values),
        fitness_values.shape
    )

    current_best_x = cells[current_index]
    current_best_fitness = fitness(current_best_x)

    # Update global best
    if current_best_fitness < best_fitness:
        best_x = current_best_x
        best_fitness = current_best_fitness

# Display results
print("Parallel Cellular Algorithm Results")
print("------------------------------------")
print("Best value of x:", best_x)
print("Minimum fitness:", best_fitness)
print("Number of cells:", GRID_SIZE * GRID_SIZE)
print("Iterations:", ITERATIONS)
```

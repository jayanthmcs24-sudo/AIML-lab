```python

from itertools import product

# Knowledge Base: P -> Q, P
# Query: Q
print("JAYANTH M-1WN24CS122")
def implies(p, q):
    return not p or q

print("Knowledge Base: P -> Q, P")
print("Query: Q\n")

print("P\tQ\tP->Q\tKB\tQuery")

entails = True

for P, Q in product([True, False], repeat=2):
    rule = implies(P, Q)
    KB = rule and P

    print(P, Q, rule, KB, Q, sep="\t")

    if KB and not Q:
        entails = False

if entails:
    print("\nKB entails Query")
else:
    print("\nKB does NOT entail Query")
```

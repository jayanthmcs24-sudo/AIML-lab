```python
import heapq

def print_state(s):
    for i in range(0, 9, 3):
        print(s[i:i+3])
    print()


# Misplaced Tiles
def h(s, goal):
    return sum(s[i] != goal[i] and s[i] != 0 for i in range(9))


# Generate moves
def moves(s):
    result = []
    p = s.index(0)
    r, c = p // 3, p % 3

    for name, dr, dc in [("Up",-1,0), ("Down",1,0),
                         ("Left",0,-1), ("Right",0,1)]:

        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            q = nr * 3 + nc
            x = list(s)
            x[p], x[q] = x[q], x[p]
            result.append((tuple(x), name))

    return result


# A* Search
def astar(start, goal):
    pq = [(h(start, goal), 0, start)]
    parent = {start: None}
    move = {}
    cost = {start: 0}

    while pq:
        f, g, s = heapq.heappop(pq)

        if s == goal:
            path = []
            while parent[s] is not None:
                path.append((move[s], s))
                s = parent[s]
            return path[::-1]

        for nxt, m in moves(s):
            ng = g + 1

            if nxt not in cost or ng < cost[nxt]:
                cost[nxt] = ng
                parent[nxt] = s
                move[nxt] = m
                heapq.heappush(pq, (ng + h(nxt, goal), ng, nxt))

    return None


# Input
print("8-PUZZLE A* - MISPLACED TILES")

start = tuple(map(int, input("Enter initial state: ").split()))
goal = tuple(map(int, input("Enter goal state: ").split()))

print("\nInitial State:")
print_state(start)

print("Goal State:")
print_state(goal)

print("Initial h(n) =", h(start, goal))
print("Initial g(n) = 0")
print("Initial f(n) =", h(start, goal))


# Solve
path = astar(start, goal)

if path:
    print("\nSolution found!")
    print("Number of moves:", len(path))

    for m, s in path:
        print("Move:", m)
        print_state(s)
else:
    print("No solution found.")
```
